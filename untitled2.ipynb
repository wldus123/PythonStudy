```python
#cylinders 그룹별 horsepower의 중앙값으로 결측치 채우기
df2['horsepower'] = df2['horsepower'].fillna(
    df2.groupby('cylinders')['horsepower'].transform('median')
)

# 결측치가 잘 채워졌는지 확인 (0이 나와야 합니다)
# print(df2['horsepower'].isnull().sum())

print(df2.iloc[32:33])  # 4기통 차량 확인
print(df2.iloc[126:127])  # 6기통 차량 확인
# print(df2.at[126, 'horsepower'])  # 6기통 차량 마력 확인
```
